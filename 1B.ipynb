```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

numeric_columns = [
    "Calories",
    "Total Fat (g)",
    "Trans Fat (g)",
    "Saturated Fat (g)",
    "Sodium (mg)",
    "Total Carbohydrates (g)",
    "Cholesterol (mg)",
    "Dietary Fibre (g)",
    "Sugars (g)",
    "Protein (g)",
    "Vitamin A (% DV)",
    "Vitamin C (% DV)",
    "Calcium (% DV)",
    "Iron (% DV)",
    "Caffeine (mg)"
]

percentage_columns = [
    "Vitamin A (% DV)",
    "Vitamin C (% DV)",
    "Calcium (% DV)",
    "Iron (% DV)"
]

def prepare_data():
    df = pd.read_csv("starbucks.csv")
    df.columns = df.columns.str.strip()
    df = df.drop_duplicates()

    for col in percentage_columns:
        df[col] = df[col].astype(str).str.replace("%", "", regex=False)

    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        df[col] = df[col].fillna(df[col].median())

    encoder = LabelEncoder()

    for col in ["Beverage", "Beverage_prep", "Beverage_category"]:
        df[col] = encoder.fit_transform(df[col].astype(str))

    X = df.drop(["Beverage_category", "Beverage"], axis=1)
    y = df["Beverage_category"]

    return df, X, y

# CRISP-DM
print("CRISP-DM METHODOLOGY")

print("\nBusiness Understanding")
print("Objective: Classify Starbucks beverages based on nutritional characteristics.")

print("\nData Understanding")
df, X, y = prepare_data()
print("Dataset Shape:", df.shape)
print("Number of Columns:", len(df.columns))

print("\nData Preparation")
print("Missing values handled and categorical data encoded.")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("\nModeling")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

print("\nEvaluation")
y_pred = model.predict(X_test)
crisp_accuracy = accuracy_score(y_test, y_pred)
print("CRISP-DM Accuracy:", crisp_accuracy)

print("\nDeployment")
print("Model is ready for deployment.")

# SEMMA
print("\nSEMMA METHODOLOGY")

print("\nS - Sample")
df, X, y = prepare_data()
print("Sample Size:", len(df))

print("\nE - Explore")
print("Features:", X.shape[1])
print("Target: Beverage Category")

print("\nM - Modify")
print("Missing values handled and categorical data encoded.")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("\nM - Model")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

print("\nA - Assess")
y_pred = model.predict(X_test)
semma_accuracy = accuracy_score(y_test, y_pred)
print("SEMMA Accuracy:", semma_accuracy * 100)

# KDD
print("\nKDD METHODOLOGY")

print("\nKDD Step 1: Selection")
df, X, y = prepare_data()
print("Selected Features:", X.shape[1])

print("\nKDD Step 2: Preprocessing")
print("Missing values handled and categorical data encoded.")

print("\nKDD Step 3: Transformation")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)

print("\nKDD Step 4: Data Mining")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

print("\nKDD Step 5: Interpretation")
y_pred = model.predict(X_test)
kdd_accuracy = accuracy_score(y_test, y_pred)
print("KDD Accuracy:", kdd_accuracy * 100)

print("\nRESULT")
print("CRISP-DM Accuracy:", crisp_accuracy)
print("SEMMA Accuracy:", semma_accuracy * 100)
print("KDD Accuracy:", kdd_accuracy * 100)
print("\nCRISP-DM, SEMMA and KDD completed successfully.")
```
