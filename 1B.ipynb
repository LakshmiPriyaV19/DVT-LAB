```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# CRISP-DM
print("CRISP-DM METHODOLOGY")

# Business Understanding
print("\nBusiness Understanding")
print("Objective: Classify Starbucks beverages based on their nutritional characteristics.")

# Data Understanding
print("\nData Understanding")
df = pd.read_csv("starbucks.csv")
df.columns = df.columns.str.strip()

print("\nFirst 5 Records")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nMissing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe())

# Data Preparation
print("\nData Preparation")
df = df.drop_duplicates()

numeric_columns = [
    "Calories",
    "Total Fat (g)",
    "Trans Fat (g)",
    "Saturated Fat (g)",
    "Sodium (mg)",
    "Total Carbohydrates (g)",
    "Cholesterol (mg)",
    "Dietary Fibre (g)",
    "Sugars (g)",
    "Protein (g)",
    "Vitamin A (% DV)",
    "Vitamin C (% DV)",
    "Calcium (% DV)",
    "Iron (% DV)",
    "Caffeine (mg)"
]

percentage_columns = [
    "Vitamin A (% DV)",
    "Vitamin C (% DV)",
    "Calcium (% DV)",
    "Iron (% DV)"
]

for col in percentage_columns:
    df[col] = df[col].astype(str).str.replace("%", "", regex=False)

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

encoder = LabelEncoder()

for col in ["Beverage", "Beverage_prep", "Beverage_category"]:
    df[col] = encoder.fit_transform(df[col].astype(str))

X = df.drop(["Beverage_category", "Beverage"], axis=1)
y = df["Beverage_category"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Modeling
print("\nModeling")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Evaluation
print("\nEvaluation")
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("CRISP-DM Accuracy:", accuracy)

# Deployment
print("\nDeployment")
print("Model is ready for deployment.")

# SEMMA
print("\nSEMMA METHODOLOGY")

# Sample
print("\nS - Sample")
df = pd.read_csv("starbucks.csv")
df.columns = df.columns.str.strip()
print(df.head())

# Explore
print("\nE - Explore")
print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())

# Modify
print("\nM - Modify")
df = df.drop_duplicates()

for col in percentage_columns:
    df[col] = df[col].astype(str).str.replace("%", "", regex=False)

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

for col in ["Beverage", "Beverage_prep", "Beverage_category"]:
    df[col] = encoder.fit_transform(df[col].astype(str))

X = df.drop(["Beverage_category", "Beverage"], axis=1)
y = df["Beverage_category"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Model
print("\nM - Model")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Assess
print("\nA - Assess")
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("SEMMA Accuracy:", accuracy * 100)

# KDD
print("\nKDD METHODOLOGY")

# Selection
print("\nKDD Step 1: Selection")
df = pd.read_csv("starbucks.csv")
df.columns = df.columns.str.strip()

# Preprocessing
print("\nKDD Step 2: Preprocessing")
df = df.drop_duplicates()

for col in percentage_columns:
    df[col] = df[col].astype(str).str.replace("%", "", regex=False)

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

for col in ["Beverage", "Beverage_prep", "Beverage_category"]:
    df[col] = encoder.fit_transform(df[col].astype(str))

X = df.drop(["Beverage_category", "Beverage"], axis=1)
y = df["Beverage_category"]

# Transformation
print("\nKDD Step 3: Transformation")
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Data Mining
print("\nKDD Step 4: Data Mining")
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

# Interpretation
print("\nKDD Step 5: Interpretation")
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("KDD Accuracy:", accuracy * 100)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nCRISP-DM, SEMMA and KDD Analysis Completed Successfully.")
```
