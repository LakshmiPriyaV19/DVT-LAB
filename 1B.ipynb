```python
import pandas as pd

# Import CSV file
df = pd.read_csv("starbucks.csv")

# Display first five records
print("First 5 Records:")
print(df.head())

# Display dataset shape
print("\nDataset Shape:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Export to CSV
df.to_csv("starbucks_processed.csv", index=False)

# Export to Excel
df.to_excel("starbucks_processed.xlsx", index=False)

# Import the exported CSV file
new_df = pd.read_csv("starbucks_processed.csv")

# Display exported data
print("\nExported Data:")
print(new_df.head())

print("\nData exported successfully.")
```
