```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("starbucks.csv")
df.columns = df.columns.str.strip()
```

```python
print(df)
```

```python
print(df.head())
# Display shape
print(df.shape)
# Display data types
print(df.dtypes)
# Display statistical summary
print(df.describe())
```

```python
print(df[df["Calories"] > 200])
```

```python
df["Caffeine (mg)"] = pd.to_numeric(
    df["Caffeine (mg)"],
    errors="coerce"
)
```

```python
print(df.sort_values("Calories", ascending=False))
```

```python
# Convert Caffeine to numeric
df["Caffeine (mg)"] = pd.to_numeric(
    df["Caffeine (mg)"],
    errors="coerce"
)

# Fill missing caffeine values with median
df["Caffeine (mg)"] = df["Caffeine (mg)"].fillna(
    df["Caffeine (mg)"].median()
)

# Filter beverages with caffeine greater than 100
print(df[df["Caffeine (mg)"] > 100])

# Display correlation matrix
print(df[["Calories", "Caffeine (mg)"]].corr())

# Display correlation
print(
    "Correlation:",
    df["Calories"].corr(df["Caffeine (mg)"])
)
```

```python
print(df.sort_values("Caffeine (mg)", ascending=False))
```

```python
print(df.groupby("Beverage_category")["Calories"].mean())
```

```python
print(df.groupby("Beverage_category")["Sugars (g)"].mean())
```

```python
print("Mean Calories:", df["Calories"].mean())

# Calculate median
print("Median Calories:", df["Calories"].median())

# Calculate minimum
print("Minimum Calories:", df["Calories"].min())

# Calculate maximum
print("Maximum Calories:", df["Calories"].max())

# Calculate standard deviation
print("Standard Deviation:", df["Calories"].std())

# Display correlation matrix
print(df[["Calories", "Caffeine (mg)"]].corr())

# Display correlation
print("Correlation:", df["Calories"].corr(df["Caffeine (mg)"]))
```

```python
plt.bar(df["Beverage_category"], df["Calories"])
plt.title("Calories by Beverage Category")
plt.xlabel("Beverage Category")
plt.ylabel("Calories")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.plot(df["Beverage_category"], df["Calories"], marker="o")
plt.title("Calories of Starbucks Beverages")
plt.xlabel("Beverage Category")
plt.ylabel("Calories")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["Calories"], bins=10)
plt.title("Calories Distribution")
plt.xlabel("Calories")
plt.ylabel("Number of Beverages")
plt.tight_layout()
plt.show()
```

```python
plt.boxplot(df["Calories"])
plt.title("Calories Box Plot")
plt.ylabel("Calories")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
plt.scatter(
    df["Caffeine (mg)"],
    df["Calories"]
)

plt.title("Calories vs Caffeine")
plt.xlabel("Caffeine (mg)")
plt.ylabel("Calories")
plt.tight_layout()
plt.show()
```
