```python
import pandas as pd
import sqlite3
import os

# Load Dataset
df = pd.read_csv("starbucks.csv")

# Display Dataset
print(df)

# First 5 rows
print(df.head())

# Last 5 rows
print(df.tail())

# Dataset Information
df.info()

# Shape
print("Shape:", df.shape)

# Column Names
print("Columns:")
print(df.columns)

# Data Types
print("Data Types:")
print(df.dtypes)

# Statistical Summary
print(df.describe())

# Summary including object columns
print(df.describe(include='all'))

# Missing Values
print("Missing Values:")
print(df.isnull().sum())

# Duplicate Rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove Duplicates (Optional)
df = df.drop_duplicates()

# Value Counts for every object column
for column in df.select_dtypes(include='object').columns:
    print("\nValue Counts for", column)
    print(df[column].value_counts())

# Unique values for every object column
for column in df.select_dtypes(include='object').columns:
    print("\nUnique values in", column)
    print(df[column].unique())

# Correlation Matrix
print(df.corr(numeric_only=True))

# Export Files
df.to_csv("starbucks_clean.csv", index=False)
df.to_excel("starbucks.xlsx", index=False)
df.to_json("starbucks.json", orient="records", indent=4)

print("CSV, Excel and JSON files exported successfully.")

# Read Excel File
df_excel = pd.read_excel("starbucks.xlsx")
print(df_excel.head())

# SQLite Database
conn = sqlite3.connect("starbucks.db")

# Store DataFrame into SQL
df.to_sql("starbucks", conn, if_exists="replace", index=False)

# First 5 Records
print(pd.read_sql("SELECT * FROM starbucks LIMIT 5", conn))

# Total Records
print(pd.read_sql("""
SELECT COUNT(*) AS Total_Records
FROM starbucks
""", conn))
```

```python
import pandas as pd
import sqlite3

# Load Dataset
df = pd.read_csv("starbucks.csv")

# Display Dataset
print(df.head())

# Dataset Information
print("\nShape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary:")
print(df.describe(include="all"))

# Create SQLite Database
conn = sqlite3.connect("starbucks.db")

# Store DataFrame
df.to_sql("starbucks", conn, if_exists="replace", index=False)

# First 5 Records
print("\nFirst 5 Records")
print(pd.read_sql("SELECT * FROM starbucks LIMIT 5", conn))

# Total Records
print("\nTotal Records")
print(pd.read_sql("SELECT COUNT(*) AS Total_Records FROM starbucks", conn))

# Get Column Names
columns = df.columns.tolist()

# Find text columns
text_cols = df.select_dtypes(include="object").columns.tolist()

print("\nText Columns:", text_cols)

# Value Counts using SQL for every text column
for col in text_cols:
    print(f"\nTop values in {col}")
    query = f'''
    SELECT "{col}", COUNT(*) AS Total
    FROM starbucks
    GROUP BY "{col}"
    ORDER BY Total DESC
    LIMIT 10
    '''
    print(pd.read_sql(query, conn))

# Find numeric columns
num_cols = df.select_dtypes(include=["int64","float64"]).columns.tolist()

print("\nNumeric Columns:", num_cols)

# Average of numeric columns
for col in num_cols:
    query = f'''
    SELECT AVG("{col}") AS Average_{col}
    FROM starbucks
    '''
    print(pd.read_sql(query, conn))

# Export Files
df.to_csv("starbucks.csv", index=False)
df.to_excel("starbucks.xlsx", index=False)
df.to_json("starbucks.json", orient="records", indent=4)

print("\nCSV, Excel and JSON exported successfully.")

conn.close()
```
