```python
import pandas as pd

# Load the Starbucks dataset
df = pd.read_csv("starbucks.csv")

# Display the complete DataFrame
print("Original DataFrame:")
print(df)

# First 5 rows
print("\n1. First 5 rows (head):")
print(df.head())

# Last 5 rows
print("\n2. Last 5 rows (tail):")
print(df.tail())

# Shape of the dataset
print("\n3. Shape (rows, columns):")
print(df.shape)

# Column names
print("\n4. Column Names:")
print(df.columns)

# Dataset information
print("\n5. DataFrame Information:")
df.info()

# Statistical summary
print("\n6. Statistical Summary:")
print(df.describe(include="all"))

# Missing values
print("\n7. Missing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\n8. Duplicate Rows:")
print(df.duplicated().sum())

# Data types
print("\n9. Data Types:")
print(df.dtypes)

# Display a particular column (replace with any column from your dataset)
print("\n10. First Column:")
print(df[df.columns[0]])

# Unique values in the first object column
object_cols = df.select_dtypes(include="object").columns

if len(object_cols) > 0:
    print(f"\n11. Unique values in '{object_cols[0]}':")
    print(df[object_cols[0]].unique())

    print(f"\n12. Value Counts of '{object_cols[0]}':")
    print(df[object_cols[0]].value_counts())

# Correlation of numeric columns
print("\n13. Correlation Matrix:")
print(df.corr(numeric_only=True))
```
