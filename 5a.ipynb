```python
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("sales_data.csv")

months = df["Month"]
sales = df["Sales"]

# Program 1: Line Chart
plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()

# Program 2: Bar Chart
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()

# Program 3: Pie Chart
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()

# Program 4: Histogram
plt.hist(sales, bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

# Program 5: Scatter Plot
plt.scatter(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()

# Program 6: Box Plot
plt.boxplot(sales)
plt.title("Sales Box Plot")
plt.ylabel("Sales")
plt.show()

# Program 7: Multiple Line Charts
profit = df["Profit"]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.title("Sales and Profit")
plt.xlabel("Months")
plt.ylabel("Value")
plt.legend()
plt.show()
```
