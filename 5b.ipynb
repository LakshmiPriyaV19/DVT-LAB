```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("starbucks.csv")
df.columns = df.columns.str.strip()

# Display first five records
print(df.head())

# Convert Caffeine column to numeric
df["Caffeine (mg)"] = pd.to_numeric(
    df["Caffeine (mg)"],
    errors="coerce"
)

df["Caffeine (mg)"] = df["Caffeine (mg)"].fillna(
    df["Caffeine (mg)"].median()
)
```

```python
plt.figure(figsize=(10, 5))
plt.bar(
    df["Beverage_category"],
    df["Calories"]
)

plt.title("Calories by Beverage Category")
plt.xlabel("Beverage Category")
plt.ylabel("Calories")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Line chart
plt.figure(figsize=(10, 5))
plt.plot(
    df["Calories"],
    marker="o"
)

plt.title("Calories of Starbucks Beverages")
plt.xlabel("Beverage Index")
plt.ylabel("Calories")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(
    df["Caffeine (mg)"],
    df["Calories"]
)

plt.title("Calories vs Caffeine")
plt.xlabel("Caffeine (mg)")
plt.ylabel("Calories")
plt.tight_layout()
plt.show()
```

```python
# Histogram
plt.figure(figsize=(8, 5))
plt.hist(
    df["Calories"],
    bins=10
)

plt.title("Calories Distribution")
plt.xlabel("Calories")
plt.ylabel("Number of Beverages")
plt.tight_layout()
plt.show()
```

```python
# Box plot
plt.figure(figsize=(6, 5))
plt.boxplot(
    df["Calories"]
)

plt.title("Calories Box Plot")
plt.ylabel("Calories")
plt.tight_layout()
plt.show()
```
