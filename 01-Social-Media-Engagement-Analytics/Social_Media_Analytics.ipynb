```python
from sqlalchemy import create_engine
username = "root"
password = "Praba_1910"
host = "localhost"
port = 3306
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/social_media_analytics"
)
print("MySQL connection created successfully!")
```
