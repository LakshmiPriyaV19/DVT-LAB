```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Lakshmi\Downloads\world_tourism_economy_data.csv")
print("Data loaded successfully!")
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
numeric_columns = df.select_dtypes(include='number').columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values handled successfully!")
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
df["revenue_per_tourist"] = (
    df["tourism_receipts"] / df["tourism_arrivals"]
)

print("New column created successfully!")
df[["country", "year", "tourism_receipts", 
    "tourism_arrivals", "revenue_per_tourist"]].head()
```

```python
df = df.sort_values(["country", "year"])

df["arrival_growth_rate"] = (
    df.groupby("country")["tourism_arrivals"]
      .pct_change() * 100
)

print("Arrival growth rate created successfully!")
df[["country", "year", "tourism_arrivals", 
    "arrival_growth_rate"]].head(10)
```

```python
df = df.drop(columns=["country_code"])

print("Unnecessary column removed successfully!")
print(df.columns)
```

```python
print("Final dataset shape:", df.shape)
print("\nFinal columns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
df.to_csv("tourism_analytics_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
